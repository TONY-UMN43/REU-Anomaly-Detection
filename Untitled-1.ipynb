```python
# ==========================================
# STEP 1: SPATIO-TEMPORAL DATA PIPELINE
# ==========================================

def create_graph_snapshots(data, edge_timestamps, num_snapshots):
    """
    Groups a massive evolving graph into a sequence of discrete-time static snapshots.
    """
    timestamps = edge_timestamps.cpu().numpy() if isinstance(edge_timestamps, torch.Tensor) else np.array(edge_timestamps)
    edge_index = data.edge_index.cpu().numpy()
    
    print(edge_index[0])
    df_edges = pd.DataFrame({
        'source': edge_index[0],
        'target': edge_index[1],
        'timestamp': timestamps
    })
    
    min_time = df_edges['timestamp'].min()
    max_time = df_edges['timestamp'].max()
    
    bins = np.linspace(min_time, max_time, num_snapshots + 1)
    df_edges['snapshot_idx'] = pd.cut(df_edges['timestamp'], bins=bins, labels=False, include_lowest=True)
    
    snapshots = []
    global_x = data.x  # Fixed shape: [num_nodes, node_features]
    
    for t in range(num_snapshots):
        snapshot_df = df_edges[df_edges['snapshot_idx'] == t]
        
        if not snapshot_df.empty:
            snap_edge_index = torch.tensor(
                [snapshot_df['source'].values, snapshot_df['target'].values], 
                dtype=torch.long
            )
        else:
            snap_edge_index = torch.empty((2, 0), dtype=torch.long)
            
        snapshot_graph = Data(x=global_x, edge_index=snap_edge_index)
        snapshots.append(snapshot_graph)
        
    return snapshots

```

```python


# ==========================================
# STEP 2: MODEL ARCHITECTURE
# ==========================================
# ==========================================
# STEP 2: MODEL ARCHITECTURE
# ==========================================

class SpatioTemporalEncoderDecoder(nn.Module):
    def __init__(self, node_in_features, spatial_hidden_dim, temporal_hidden_dim, num_classes=1):
        super(SpatioTemporalEncoderDecoder, self).__init__()
        
        # 1. Spatial Encoder: Learns structural layout per snapshot
        self.gnn_conv1 = SAGEConv(node_in_features, spatial_hidden_dim)
        self.gnn_conv2 = SAGEConv(spatial_hidden_dim, spatial_hidden_dim)
        
        # 2. Temporal Encoder: Tracks how spatial embeddings evolve over snapshots
        self.temporal_rnn = nn.GRU(
            input_size=spatial_hidden_dim,
            hidden_size=temporal_hidden_dim,
            num_layers=1,
            batch_first=True  # Expected input shape: [num_nodes, num_snapshots, spatial_hidden_dim]
        )
        
        # 3. Decoder: Map the final spatio-temporal state to the target prediction
        self.decoder = nn.Sequential(
            nn.Linear(temporal_hidden_dim, temporal_hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(temporal_hidden_dim // 2, num_classes)
        )
        
    # FIX: Indented this block so it is a method of SpatioTemporalEncoderDecoder
    def forward(self, snapshot_list, full_edge_index):
        """
        snapshot_list: List of PyG Data objects for the spatial loops
        full_edge_index: The global edge index [2, 4554343] to compute predictions for
        """
        num_nodes = snapshot_list[0].x.size(0)
        num_snapshots = len(snapshot_list)
        spatial_dim = self.gnn_conv2.out_channels
        
        spatial_sequence = torch.zeros(num_nodes, num_snapshots, spatial_dim, device=snapshot_list[0].x.device)
        
        # ---- 1. Spatial Loop ----
        for t, snapshot in enumerate(snapshot_list):
            x, edge_index = snapshot.x, snapshot.edge_index
            h = self.gnn_conv1(x, edge_index)
            h = F.relu(h)
            h = self.gnn_conv2(h, edge_index)
            h = F.relu(h)
            spatial_sequence[:, t, :] = h
            
        # ---- 2. Temporal Loop ----
        rnn_out, _ = self.temporal_rnn(spatial_sequence)
        final_node_embeddings = rnn_out[:, -1, :]  # Shape: [num_nodes, temporal_hidden_dim]
        
        # ---- 3. Edge Decoder Loop ----
        src_nodes, dst_nodes = full_edge_index[0], full_edge_index[1]
        
        src_embeddings = final_node_embeddings[src_nodes]  
        dst_embeddings = final_node_embeddings[dst_nodes]  
        
        edge_representations = src_embeddings * dst_embeddings  
        edge_predictions = self.decoder(edge_representations)  # Shape: [4554343, 1]
        
        return edge_predictions



# ==========================================
# HOW TO RUN EXECUTION EXAMPLE
# ==========================================
# 1. Build snapshots from your raw 4.5M edge dataset
# snapshots_sequence = create_graph_snapshots(data_load, data_load.edge_time, num_snapshots=12)
# 
# 2. Initialize Model
# model = SpatioTemporalEncoderDecoder(
#     node_in_features=data_load.x.size(1), 
#     spatial_hidden_dim=32, 
#     temporal_hidden_dim=64, 
#     num_classes=1
# )
# 
# 3. Run training using your node masks and targets
# train_spatio_temporal_model(model, snapshots_sequence, data_load.train_node_mask, data_load.y_node, epochs=20)
```

```python
def train_spatio_temporal_model(model, snapshots, full_edge_index, train_edge_mask, edge_labels, epochs, lr=0.005):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.BCEWithLogitsLoss() 
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    snapshots = [snap.to(device) for snap in snapshots]
    full_edge_index = full_edge_index.to(device)
    train_edge_mask = train_edge_mask.to(device)
    edge_labels = edge_labels.to(device).float()
    
    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        
        # Pass both the snapshots AND the global edge layout to the model
        output = model(snapshots, full_edge_index) # Drops shape to [4554343]
        
        # Calculate loss using edge mask against edge labels
        loss = loss_fn(output[train_edge_mask], edge_labels[train_edge_mask])
        
        loss.backward()
        optimizer.step()
        
        print(f"Epoch {epoch+1:02d} | Loss: {loss.item():.4f}")
```

```python
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
@torch.no_grad()
def evaluate_spatio_temporal_model(model, snapshots, full_edge_index, test_edge_mask, edge_labels, batch_size=250000):
    """
    Evaluates the model by chunking test edges into mini-batches to prevent memory crashes.
    """
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.eval()
    
    # 1. Get the final temporal node embeddings once [num_nodes, temporal_hidden_dim]
    num_nodes = snapshots[0].x.size(0)
    num_snapshots = len(snapshots)
    spatial_dim = model.gnn_conv2.out_channels
    
    spatial_sequence = torch.zeros(num_nodes, num_snapshots, spatial_dim, device=device)
    for t, snapshot in enumerate(snapshots):
        x, edge_index = snapshot.x, snapshot.edge_index
        h = model.gnn_conv1(x, edge_index)
        h = F.relu(h)
        h = model.gnn_conv2(h, edge_index)
        h = F.relu(h)
        spatial_sequence[:, t, :] = h
        
    rnn_out, _ = model.temporal_rnn(spatial_sequence)
    final_node_embeddings = rnn_out[:, -1, :]  # Shape: [25525, hidden_dim]
    
    # 2. Extract only the edge pairs that belong to the test set
    test_indices = torch.where(test_edge_mask)[0]
    test_src = full_edge_index[0, test_indices].to(device)
    test_dst = full_edge_index[1, test_indices].to(device)
    
    # Cast true labels safely
    test_true_tensor = edge_labels[test_edge_mask].to(device).squeeze(-1)
    
    # 3. Process test edges in safe chunks
    all_test_probs = []
    num_test_edges = test_indices.size(0)
    
    for i in range(0, num_test_edges, batch_size):
        batch_src = test_src[i:i+batch_size]
        batch_dst = test_dst[i:i+batch_size]
        
        # Look up embeddings just for this chunk
        src_emb = final_node_embeddings[batch_src]
        dst_emb = final_node_embeddings[batch_dst]
        
        edge_rep = src_emb * dst_emb
        batch_logits = model.decoder(edge_rep).squeeze(-1)
        batch_probs = torch.sigmoid(batch_logits)
        
        all_test_probs.append(batch_probs.cpu())
        
    # Combine chunked predictions
    test_probs = torch.cat(all_test_probs).numpy()
    test_true = test_true_tensor.cpu().numpy()
    anomaly_ratio = (test_true == 1).mean()
    threshold = np.percentile(test_probs, (1 - anomaly_ratio) * 100)

    print(f"Using dynamically adjusted threshold: {threshold:.4f}")
    test_preds = (test_probs >= threshold).astype(int)    
    # 4. Compute Metrics
    auc = roc_auc_score(test_true, test_probs)
    precision = precision_score(test_true, test_preds, zero_division=0)
    recall = recall_score(test_true, test_preds, zero_division=0)
    f1 = f1_score(test_true, test_preds, zero_division=0)
    accuracy = (test_preds == test_true).mean()
    classes = np.array([0,1])
    test_matrix = confusion_matrix(test_true,test_preds,labels=classes)
    plt.figure(figsize=(6, 5))
    sns.heatmap(test_matrix, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Normal', 'Anomaly'], 
                yticklabels=['Normal', 'Anomaly'])
    plt.ylabel('Actual Labels')
    plt.xlabel('Predicted Labels')
    plt.title('DARPA Dataset Anomaly Confusion Matrix')
    plt.show()    
    # print(test_matrix)
    print("\n================ TEST PERFORMANCE ================")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"AUC-ROC:   {auc:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    print("==================================================")
    
    return {"auc": auc, "f1": f1, "precision": precision, "recall": recall}
```
